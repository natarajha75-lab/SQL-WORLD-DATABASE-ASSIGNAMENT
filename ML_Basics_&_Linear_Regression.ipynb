###Q3. What is Linear Regression? Write the Linear Regression equation and explain intercept and coefficient.

# Linear Regression:- A supervised learning regression algorithm that models the linear relationship between a dependent target variable (y) and one or more independent predictor variables(x).
* Equation (Simple Liner Regresssion):
(y=β₀+β₁x+ε)
* Explanation of Terms:
* Ŷ : The dependent variable (target value we want to predict).
* x : The independent variable(input feature).
* b₀(intercept): The expected value of (y) when all indepenent variables(x) are zero. it represents where the regression line crosses the vertical y-axis.
* b₁(Coefficient/slope):Represents the average change in (y) for a one unit change in the predictor variable(x).
* ε:The error term (residual) represnting the fifference between actual and predicted values.

### Q4. Explain the following evaluation metrics used in Linear Regression:

* SSE
* MSE
* RMSE
* R² Score
* Adjusted R²




* SSE(Sum of Squared Errors): Measures the total squared deviation of the actual values from the predicted regression line.
$$SSE = \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$$

* MSE(Mean Squared Error): The average of the squared difference between the actual and predicted values, penalizing larger errors more heavily.
$$MSE = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$$

* RMSE(Root Mean Squared Error): The square root of the MSE, wich brings the errors metric back into the original units of the target variable for easier interpretation.
$$RMSE = \sqrt{\frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2}$$

* R²Score(Coeffcient of Determination):Indicates the propration of variance  in the depenent variable that is predictable from the independent varibles (ranges from 0 to 1, or negative for arbitrarily worse models).
R²Score = $1 - \frac{\sum(Y_i - \hat{Y}_i)^2}{\sum(Y_i - \bar{Y})^2}$)

* Adjusted R²: A modified version of R² that penalizes the model for adding irrelevant features, adjusting for the number of predictors(p) and sample size(n).
$$\text{Adjusted } R^2 = 1 - \left[ \frac{(1 - R^2)(n - 1)}{n - p - 1} \right]$$
