Data Quality and Validation in ETL

# Question 1 : Define Data Quality in the context of ETL pipelines. Why is it more than just data cleaning?

## Answer:
Data Quality in ETL pipelines means ensuring that data is accurate, complete, consistent, valid, and reliable throughout the Extract, Transform, and Load process.

It is more than just data cleaning because data cleaning only fixes errors such as missing or duplicate values, while data quality also includes validating, monitoring, and maintaining the data so it is suitable for analysis and decision-making.

# Question 2 : Explain why poor data quality leads to misleading dashboards and incorrect decisions.

## Answer:
Poor data quality leads to misleading dashboards and incorrect decisions because inaccurate, incomplete, or inconsistent data produces unreliable reports and analysis. As a result, businesses may make wrong decisions based on incorrect information.

# Question 3 : What is duplicate data? Explain three causes in ETL pipelines

## Answer:
Duplicate data is the presence of the same record more than once in a dataset.

Three causes of duplicate data in ETL pipelines are:

1) Multiple data sources containing the same records.
2) Repeated data loading due to ETL process errors.
3) Lack of unique keys to identify and prevent duplicate records.

## Question 4 : Differentiate between exact, partial, and fuzzy duplicates

## Answer:
1) Exact duplicates: Records are completely identical.
2) Partial duplicates: Some fields are the same, while others are different.
3) Fuzzy duplicates: Records are similar but not exactly the same due to        -  spelling or formatting differences.

# Question 5 : Why should data validation be performed during transformation rather than after loading?

### Answer:
Data validation should be performed during the transformation stage because it identifies and corrects errors before the data is loaded into the target system. This helps ensure that only accurate, consistent, and reliable data is stored, reducing the need for later corrections.

# Question 6 : Explain how business rules help in validating data accuracy. Give an example.

### Answer:
Business rules validate data by checking whether it meets predefined conditions. This helps ensure the data is accurate and consistent.

Example: A customer's Age must be greater than 0, or an employee's Salary cannot be negative.

# Question 7 : Write an SQL query on SALES_TRANSACTION to list all duplicate keys and their counts using the
business key (Customer_ID + Product_ID + Txn_Date + Txn_Amount )

## SQL QUERY

```python
SELECT Customer_ID,
       Product_ID,
       Txn_Date,
       Txn_Amount,
       COUNT(*) AS Duplicate_Count
FROM Sales_Transactions
GROUP BY Customer_ID, Product_ID, Txn_Date, Txn_Amount
HAVING COUNT(*) > 1;
```

# Question 8 : Enforcing Referential Integrity

## SQL QUERY

```python
select s. Customer_ID
From Sales_Transactions s
Left join Customers_Master c
On s. Customer_ID = c. customerID
Where c. customerID is NULL;
```
