##Question 1 :  Write a Python program to check whether a number is even or odd using conditional statements.

```python
num=int(input("Enter a number: "))
if (num % 2) == 0:
   print("{0} is Even".format(num))
else:
   print("{0} is Odd".format(num))
```

###Question 4 : Create a new column called Revenue, where:

Revenue = Quantity × Price

Display the updated dataset with the new column.

```python
import pandas as pd
data = {'Quantity': [10, 5, 8, 12],
        'Price': [2.5, 10.0, 4.0, 1.5],
        'Region': ['South', 'North', 'East', 'South']}
df = pd.DataFrame(data)

df['Revenue'] = df['Quantity'] * df['Price']
print(df)
```
